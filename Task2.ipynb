#Data Cleaning and Feature Engineering
Fill missing values and create new features like Family Size and Age Group.

```python
#Check for and fill missing values

print("Missing Value Counts Before Cleaning:")
print(data.isnull().sum())

# Fill missing values
data['Survived'] = data['Survived'].fillna(data['Survived'].mode()[0])
data['Age'] = data['Age'].fillna(data['Age'].mean())
data['Fare'] = data['Fare'].fillna(data['Fare'].mean())
data['Cabin'] = data['Cabin'].fillna(data['Cabin'].mode()[0])
data['Embarked'] = data['Embarked'].fillna(data['Embarked'].mode()[0])

print("Missing Value Counts After Cleaning:")
print(data.isnull().sum())
```

```python
# Feature Engineering

# Family Size
data['Family_size'] = data['SibSp'] + data['Parch'] + 1

# AgeGroup Binning
bins = [0, 12, 19, 35, 60, 100]
labels = ['Child', 'Teen', 'Young Adult', 'Adult', 'Senior']
data['AgeGroup'] = pd.cut(data['Age'], bins=bins, labels=labels)

print(data[['Family_size', 'Age', 'AgeGroup']].head())
```

#Survival Count by Family Type

```python
#Survival Count by Family Type
def family_type(row):
    if row['Family_size'] == 1:
        return 'Single'
    elif row['Family_size'] <= 4:
        return 'Small Family'
    else:
        return 'Large Family'

data['Family_type'] = data.apply(family_type, axis=1)

family_type_survival = data.groupby(['Family_type','Survived']).size().reset_index(name='Count')
fig = px.bar(
    family_type_survival,
    x='Family_type', y='Count', color='Survived', barmode='group',
    color_discrete_sequence=["indianred","mediumseagreen"],
    title='Survival Count by Family Type',
    labels={'Family_type':'Family Type','Count':'Passenger Count','Survived':'Survived (0 = No, 1 = Yes)'}
)
fig.update_layout(width=800, height=600) # Set figure size for Plotly
fig.show()
```

#Sex Distribution for Most Sold Ticket

```python
#Sex Distribution for Most Sold Ticket
most_common_ticket = data['Ticket'].value_counts().idxmax() # Calculate most common ticket
print("Most sold ticket:", most_common_ticket) # Print most common ticket
print() # Print a newline for better formatting

filtered_passangers = data[data['Ticket'] == most_common_ticket]
sex_counts_most_ticket = filtered_passangers['Sex'].value_counts().reset_index()
sex_counts_most_ticket.columns=['Sex','Count']

fig = px.bar(
    sex_counts_most_ticket,
    x='Sex', y='Count', color='Sex',
    color_discrete_map={'male':'teal','female':'salmon'}, # Changed colors
    title=f'Sex Distribution for Most Sold Ticket: {most_common_ticket}',
    labels={'Count':'Passenger Count'}
)
fig.update_layout(width=800, height=600) # Set figure size for Plotly
fig.show()
```

#Sex Distribution for least Sold Ticket

```python
#Sex Distribution for least Sold Ticket
least_sold_ticket = data['Ticket'].value_counts().idxmin()
print("Least sold ticket:", least_sold_ticket)
print()

filtered_passengers = data[data['Ticket'] == least_sold_ticket]
print(f"Number of passengers with the least sold ticket ({least_sold_ticket}): {len(filtered_passengers)}") # Add count of passengers

sex_counts = filtered_passengers['Sex'].value_counts().reset_index()
sex_counts.columns = ['Sex', 'Count']

# Ensure both male and female are in the sex_counts for plotting and printing
all_sexes = pd.DataFrame({'Sex': ['male', 'female'], 'Count': 0})
sex_counts = pd.merge(all_sexes, sex_counts, on='Sex', how='left', suffixes=('_all', '_filtered')).fillna(0)
sex_counts['Count'] = sex_counts['Count_all'] + sex_counts['Count_filtered']
sex_counts = sex_counts[['Sex', 'Count']]


print(sex_counts)

import plotly.express as px

fig = px.bar(sex_counts,
             x='Sex',
             y='Count',
             color='Sex',
             title=f'Sex Distribution for Least Sold Ticket: {least_sold_ticket}') # Updated title
fig.update_layout(width=600, height=500) # Set figure size for Plotly
fig.show()
```
